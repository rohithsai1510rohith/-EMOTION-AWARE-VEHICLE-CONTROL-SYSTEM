```python
!pip install -q chromadb sentence-transformers

import chromadb
from chromadb.utils import embedding_functions

client = chromadb.Client()
print("ChromaDB ready!")
```

```python
collection = client.create_collection(
    name="my_own_collection"
)

sentences = [
    "Python is useful for artificial intelligence.",
    "Java is an object-oriented programming language.",
    "Machine learning helps computers learn from data.",
    "Football is a popular outdoor sport.",
    "Deep learning uses neural networks."
]

collection.add(
    documents=sentences,
    ids=["id1", "id2", "id3", "id4", "id5"]
)

results = collection.query(
    query_texts=["How do computers learn?"],
    n_results=3
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
category_collection = client.create_collection(
    name="category_collection"
)

documents = [
    "Python is used in AI development.",
    "Java is used for application development.",
    "Football is played by two teams.",
    "Cricket is a popular outdoor game.",
    "Machine learning analyzes data."
]

metadatas = [
    {"category": "technology"},
    {"category": "technology"},
    {"category": "sports"},
    {"category": "sports"},
    {"category": "technology"}
]

ids = ["c1", "c2", "c3", "c4", "c5"]

category_collection.add(
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

# Query technology category
print("TECHNOLOGY")
results = category_collection.query(
    query_texts=["Programming and AI"],
    n_results=3,
    where={"category": "technology"}
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")

# Query sports category
print("\nSPORTS")
results = category_collection.query(
    query_texts=["Outdoor games"],
    n_results=2,
    where={"category": "sports"}
)

for doc, distance in zip(
    results["documents"][0],
    results["distances"][0]
):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
# Embedding models
mini_model = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

mpnet_model = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-mpnet-base-v2"
)

docs = [
    "I love dogs and puppies.",
    "Machine learning is useful for AI.",
    "Football is my favorite sport."
]

query = ["I like pets"]

# MiniLM collection
mini_collection = client.create_collection(
    name="minilm_collection",
    embedding_function=mini_model
)

mini_collection.add(
    documents=docs,
    ids=["m1", "m2", "m3"]
)

mini_results = mini_collection.query(
    query_texts=query,
    n_results=3
)

# MPNet collection
mpnet_collection = client.create_collection(
    name="mpnet_collection",
    embedding_function=mpnet_model
)

mpnet_collection.add(
    documents=docs,
    ids=["p1", "p2", "p3"]
)

mpnet_results = mpnet_collection.query(
    query_texts=query,
    n_results=3
)

# Compare distances
print("MiniLM Distance Scores")
for doc, distance in zip(
    mini_results["documents"][0],
    mini_results["distances"][0]
):
    print(f"{distance:.4f} | {doc}")

print("\nMPNet Distance Scores")
for doc, distance in zip(
    mpnet_results["documents"][0],
    mpnet_results["distances"][0]
):
    print(f"{distance:.4f} | {doc}")
```
