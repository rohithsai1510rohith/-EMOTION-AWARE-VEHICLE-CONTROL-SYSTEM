```python
import math

def cosine_similarity_manual(A, B):
    dot = sum([a * b for a, b in zip(A, B)])
    mag_A = math.sqrt(sum([a * a for a in A]))
    mag_B = math.sqrt(sum([b * b for b in B]))
    return dot / (mag_A * mag_B)
```

```python
A = [1, 1, 1]
B = [5, 5, 5]

result = cosine_similarity_manual(A, B)

print("Cosine Similarity:", result)
```

```python
import math

def cosine_similarity_manual(A, B):
    dot = sum(a * b for a, b in zip(A, B))
    mag_A = math.sqrt(sum(a * a for a in A))
    mag_B = math.sqrt(sum(b * b for b in B))
    return dot / (mag_A * mag_B)

vector_A = [2, 4, 1]
vector_B = [1, 3, 5]

result = cosine_similarity_manual(vector_A, vector_B)
print("Cosine Similarity:", round(result, 4))
```

```python
import numpy as np

A = np.array([1, 1, 1])
B = np.array([5, 5, 5])

result_numpy = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

print("NumPy Similarity:", round(result_numpy, 4))
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

sentence_A = "I love machine learning"
sentence_B = "I enjoy artificial intelligence"

vector_A = model.encode(sentence_A).tolist()
vector_B = model.encode(sentence_B).tolist()

print("Embedding A:", vector_A[:5])
print("Embedding B:", vector_B[:5])
```

```python
result_manual = cosine_similarity_manual(vector_A, vector_B)

print("Manual Similarity:", round(result_manual, 4))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

A = np.array(vector_A).reshape(1, -1)
B = np.array(vector_B).reshape(1, -1)

result_sklearn = cosine_similarity(A, B)[0][0]

print("Sklearn Similarity:", round(result_sklearn, 4))
```

```python
print("Manual:", round(result_manual, 6))
print("Sklearn:", round(result_sklearn, 6))

if np.isclose(result_manual, result_sklearn):
    print("Both results match!")
```
